This notebook demonstrates how to download background music from a YouTube video and convert it into a transcription using **yt-dlp** and **Whisper**.

## Step 1: Download Audio from YouTube Video

Enter the YouTube URL below:

```python
import subprocess

youtube_url = input('Enter the YouTube URL: ').strip()
if not youtube_url:
    raise ValueError('No URL provided!')

print('Downloading audio...')
# Use yt-dlp to download audio, extract it, and convert to MP3
command = [
    'yt-dlp',
    '-x',
    '--audio-format', 'mp3',
    '-o', '%(title)s.%(ext)s',
    youtube_url
]

result = subprocess.run(command)
if result.returncode != 0:
    raise Exception('yt-dlp failed to download the audio.')
print('Download completed.')
```

## Step 2: Locate the MP3 File

The following cell searches the current directory for the downloaded MP3 file.

```python
import glob

# Search for MP3 files in the current directory
mp3_files = glob.glob('*.mp3')
if not mp3_files:
    raise Exception('No MP3 file found in the current directory.')

mp3_file = mp3_files[0]
print(f'Found MP3 file: {mp3_file}')
```

## Step 3: Transcribe the Audio using Whisper

The following cell runs Whisper to convert the MP3 audio into a text transcription. Adjust the model as needed.

```python
print('Transcribing audio with Whisper...')

command = ['whisper', mp3_file, '--model', 'small']  # You can change 'small' to another model if desired
result = subprocess.run(command)

if result.returncode != 0:
    raise Exception('Whisper failed to transcribe the audio.')
print('Transcription completed. Check the generated text file for the transcription.')
```
